```python
import sqlite3

connection = sqlite3.connect(
    "/Users/akshitajoshi/Hospital_Data_Analytics_Internship/Hospital-Operations-Analytics.db"
)
import pandas as pd
import numpy as np
```

```python
patients_df = pd.read_sql_query("SELECT * FROM patients;", connection)

doctors_df = pd.read_sql_query("SELECT * FROM doctors;", connection)

departments_df = pd.read_sql_query("SELECT * FROM departments;", connection)

appointments_df = pd.read_sql_query("SELECT * FROM appointments;", connection)

encounters_df = pd.read_sql_query("SELECT * FROM encounters;", connection)

billing_df = pd.read_sql_query("SELECT * FROM billing;", connection)

payments_df = pd.read_sql_query("SELECT * FROM payments;", connection)

prescriptions_df = pd.read_sql_query("SELECT * FROM prescriptions;", connection)
```

```python
dashboard_df = appointments_df.merge(
    patients_df[
        [
            "patient_id",
            "first_name",
            "last_name",
            "gender",
            "blood_group",
            "registration_date"
        ]
    ],
    on="patient_id",
    how="left"
)

dashboard_df.head()
```

```python
dashboard_df = dashboard_df.merge(
    doctors_df[
        [
            "doctor_id",
            "first_name",
            "last_name",
            "specialization",
            "department_id"
        ]
    ],
    on="doctor_id",
    how="left",
    suffixes=("_patient", "_doctor")
)

dashboard_df.head()
```

```python
dashboard_df = dashboard_df.merge(
    departments_df[
        [
            "department_id",
            "department_name"
        ]
    ],
    on="department_id",
    how="left"
)

dashboard_df.head()
```

```python
dashboard_df = dashboard_df.merge(
    encounters_df[
        [
            "encounter_id",
            "appointment_id",
            "diagnosis",
            "consultation_duration"
        ]
    ],
    on="appointment_id",
    how="left"
)

dashboard_df.head()
```

```python
dashboard_df = dashboard_df.merge(
    billing_df[
        [
            "bill_id",
            "encounter_id",
            "total_amount"
        ]
    ],
    on="encounter_id",
    how="left"
)

dashboard_df.head()
```

```python
dashboard_df = dashboard_df.merge(
    payments_df[
        [
            "bill_id",
            "payment_status",
            "payment_method",
            "amount_paid"
        ]
    ],
    on="bill_id",
    how="left"
)

dashboard_df.head()
```

```python
dashboard_df["Patient Name"] = (
    dashboard_df["first_name_patient"] +
    " " +
    dashboard_df["last_name_patient"]
)

dashboard_df["Doctor Name"] = (
    dashboard_df["first_name_doctor"] +
    " " +
    dashboard_df["last_name_doctor"]
)
```

```python
dashboard_df = dashboard_df[
    [
        "appointment_id",
        "Patient Name",
        "gender",
        "blood_group",
        "registration_date",
        "Doctor Name",
        "specialization",
        "department_name",
        "appointment_status",
        "diagnosis",
        "consultation_duration",
        "total_amount",
        "amount_paid",
        "payment_status",
        "payment_method"
    ]
]

dashboard_df.head()
```

```python
dashboard_df.to_csv(
    "hospital_dashboard_data.csv",
    index=False
)

print("Dashboard dataset created successfully!")
```
